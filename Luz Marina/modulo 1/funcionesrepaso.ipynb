```python
#5. Calcular el interés compuesto.
#Descripción: Crea una función llamada calcular_interes_compuesto que tome el capital inicial, la tasa de interés y 
# el número de períodos como parámetros y devuelva el beneficio final después del interés compuesto.

def calcular_interes_compuesto (capital_inicial, tasa_de_interes, numero_de_periodos):
    beneficio_final = capital_inicial * ((1 + (tasa_de_interes/100))**numero_de_periodos) 
    return beneficio_final

capital_inicial = 100
tasa_de_interes = 10
numero_de_periodos = 2
calcular_interes_compuesto(capital_inicial, tasa_de_interes, numero_de_periodos)
```

```python
#6. Validar si una palabra es un anagrama de otra.
#Descripción: Crea una función llamada es_anagrama que tome dos palabras como parámetros y 
# devuelva True si una es un anagrama de la otra, o False si no lo es.

def es_anagrama(palabra1,palabra2):
    palabra1 =palabra1.lower()
    palabra2 = palabra2.lower()
    palabra1_arreglo = list(palabra1)
    palabra2_arreglo = list(palabra2)
    palabra1_arreglo.sort() 
    palabra2_arreglo.sort()
    palabra1_ordenada = "".join(palabra1_arreglo)
    palabra2_ordenada = "".join(palabra2_arreglo)
    return palabra1_ordenada == palabra2_ordenada

palabra1 = "Belen estaba cagando"
palabra2 = "Belen te vas y la cagas"
es_anagrama(palabra1,palabra2)
```

```python
def es_isograma(palabra):
    # Convertir la palabra a minúsculas
    palabra = palabra.lower()
    
    # Usar un set para verificar si hay letras repetidas
    letras_vistas = set()
    
    for letra in palabra:
        if letra.isalpha():  # Ignorar caracteres que no son letras
            if letra in letras_vistas:
                return False  # Si una letra ya fue vista, no es un isograma
            letras_vistas.add(letra)
    
    return True  # Si no se encontraron letras repetidas, es un isograma

# Ejemplos de uso:
print(es_isograma("Dermatoglyphics"))  # Output: True
print(es_isograma("aba"))             # Output: False
print(es_isograma("moOse"))           # Output: False (porque 'o' se repite)
```
