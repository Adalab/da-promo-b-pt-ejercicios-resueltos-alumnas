```python
# sacar pares de una lista
lista = [1,2,3,4,5,6,7,8,9]
lista_pares = []

def pares():
    for num in lista:
        if num % 2 == 0:
            lista_pares.append(num)
        else:
            pass
    return lista_pares

print(pares())
```
