```python
#Escribe un programa usando la funcion input()

lista_compra = []

while True:
    elemento = input("Añade un elemento. Escribe FIN para finalizar")
    
    if elemento == "FIN":
        break
    lista_compra.append(elemento)
    
    
print("\n".join(lista_compra))
```
