**Ejercicios while**

```python
#1.Calcular la suma de los primeros 10 números naturales utilizando un bucle while
# Inicializamos las variables
suma = 0
numero = 1

# Utilizamos un bucle while para sumar los primeros 10 números naturales
while numero <= 10:
    suma += numero
    numero += 1

# Imprimimos el resultado
print("La suma de los primeros 10 números naturales es:", suma)
```

```python
# 2. Juego de adivinanza utilizando un bucle while
#Descripción: Crea un juego de adivinanza en el que el jugador tiene que adivinar un número secreto 
# (tendrá que ser entre el 0 y el 10). Utiliza un bucle while para permitir múltiples intentos hasta que el jugador adivine correctamente.
import random

print("¡Bienvenida al juego de la adininaza")
print("Estoy pensando en un número entre 0 y 10")
numero_secreto = random.randint(0,10)
adivinanza = None
while adivinanza != numero_secreto:
    adivinanza = int(input("Adivina el numero:"))
    if adivinanza < numero_secreto:
         print("Ups! te has pasado, ¡Intentalo de nuevo!")
    elif adivinanza > numero_secreto:
        print("Ups! No has llegado, ¡Intentalo de nuevo!")
    else:
        print("¡Me has leido la mente!")
```

```python
# 1. Imprimir los números del 1 al 5 utilizando un bucle for
# Descripción: Utiliza un bucle for para imprimir los números del 1 al 5 (incluido) en orden ascendente.

for numero in range(1, 6):
    print(numero)
```

```python
#2. Calcular la suma de los primeros 10 números naturales utilizando un bucle for
# Descripción: Utiliza un bucle for para calcular la suma de los primeros 10 números naturales (1 + 2 + 3 + ... + 10).

suma= 0
for numero in range(1, 11):
    suma += numero
    
print(suma)
```

```python
# 3. Calcular el producto de los elementos de una lista
# Descripción: Dada una lista de números, utiliza un bucle for para calcular el producto de todos los elementos.
lista = [2, 3, 4, 5]
producto = 1

for numero in lista:
    producto *= numero

print("El producto de los elementos de la lista es:", producto)
```

```python
#4. Imprimir los caracteres de una cadena de texto
#Descripción: Dada una cadena de texto, utiliza un bucle for para imprimir cada carácter en una línea separada.
cadena = "Pepe el más guapo"
for letra in cadena:
    print(letra)
```

```python
# 5. Imprimir los números pares del 1 al 100 utilizando un bucle for
#Descripción: Utiliza un bucle for para imprimir los números pares del 1 al 100 en orden ascendente.

for numero in range(2, 101, 2):
    print(numero)
    
```

```python
#6. Imprimir los números primos del 1 al 50 utilizando un bucle for
#Descripción: Utiliza un bucle for para imprimir los números primos del 1 al 50 en orden ascendente.
#NOTA: Según la Wikipedia, un número primo es un número natural mayor que 1 que no tiene divisores 
# positivos más que 1 y él mismo.

for numero in range(1, 51):
    if numero > 1:
        es_primo = True
        for i in range(2, numero):
            if numero % i == 0:
                es_primo = False
                break
        if es_primo:
            print(numero)

    
```

```python
#7. Calcular el factorial de un número utilizando un bucle for
numero = 5
factorial = 1

for i in range(1, numero+1):
    factorial *= i

print("El factorial de", numero, "es:", factorial)
#Descripción: Dado un número entero positivo, calcula su factorial utilizando un bucle for.
```

```python
#8. Juego de adivinar palabras utilizando un bucle for
# Descripción: Crea un juego en el que el jugador debe adivinar una palabra secreta. 
# El jugador tiene un número limitado de intentos y se le proporcionan pistas después de cada intento. 
# Utiliza un bucle for para controlar los intentos del jugador.

palabra_secreta = "python"
intentos_maximos = 5

for intento in range(intentos_maximos):
    adivinanza = input("Intenta adivinar la palabra secreta: ")
    if adivinanza == palabra_secreta:
        print("¡Felicidades! Adivinaste la palabra secreta.")
        break
    else:
        print("Incorrecto. Intenta nuevamente.")

if adivinanza != palabra_secreta:
    print("Agotaste tus intentos. La palabra secreta era:", palabra_secreta)
```

## Ejercicios _list comprehension_ ##

```python
#1. Generar una lista de los números pares del 1 al 20.
# Descripción: Utiliza list comprehension para generar una lista que contenga todos los números pares en el rango 
# del 1 al 20.
pares = [numero for numero in range(1, 21) if numero % 2 == 0]
print(pares)
```

```python
#2. Ejercicio: Obtener una lista de las vocales en una cadena de texto. 
# Descripción: Utiliza list comprehension para crear una lista que contenga todas las vocales presentes en una cadena de texto (sin repetición).

cadena = "soy una pringada"
vocales = [caracter for caracter in cadena.lower() if caracter in 'aeiou']
print(vocales)
```

```python
#3. Filtrar los números positivos de una lista y calcular su cuadrado.
# Descripción: Utiliza list comprehension para filtrar los números positivos de una lista y calcular el cuadrado de cada uno.

lista= [1,2,3,4,5]
cuadrado= [num**2 for num in lista]
print(cuadrado)
```
