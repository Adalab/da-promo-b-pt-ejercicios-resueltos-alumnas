```python
import pandas as pd 
import numpy as np
```

```python
df1 = pd.read_csv("../files/rating-and-performance.csv")
df2 = pd.read_csv("../files/insight.csv")
```

```python
df3 = df1(df2, left_on='title', right_on = "title_orig")
df3.head()
```

```python
df3 = df1.merge(df2, on="id", how="left")
```

```python
#1. Unid los dos DataFrames en uno solo, usando el método que considereis más correcto. 
# Querremos quedarnos con todos los productos que estén en los dos conjuntos de datos.
df3 = pd.concat([df1,df2], axis =0, ignore_index=True)
```

```python
df3.head()
```

```python
#2.Explorad el nombre de las columnas, ¿hay alguna que esté repetida o que no nos aporte información de valor?
df3.drop(["title", "mean.product.prices","mean.discount"], axis = 1, inplace = True)
```

```python
df3.head(2)
```

```python
# ¿Hay alguna columna que solo tenga un valor? En caso de ser así, eliminadlas. 
#df3["currency_buyer"].value_counts()
columnas_un_valor = []
for column in df3.columns:
    if df3[column].value_counts().shape[0] == 1:
        columnas_un_valor.append(column)
        
df3.drop(columnas_un_valor, axis=1)
```

```python
#3.Algunas de las columnas tienen . en sus nombres y en otras tenemos _. 
# Unifica el nombre de las columnas para que todas tengan "_". Además, todas deben estar en minúsculas.
col_nuevas = {col : col.lower().replace(".", "_") for col in df3.columns}
df3.rename(columns= col_nuevas, inplace = True) 
```
