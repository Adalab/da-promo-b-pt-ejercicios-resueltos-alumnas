```python
#1.Abre el fichero de rating-and-performance con el método que sea más correcto. Además muestra las 5 primeras filas del DataFrame.

df = pd.read_csv("../files/rating-and-performance.csv",index_col= 0)
df.head()
```

```python
#2. Pongamos en práctica el loc e iloc:
#Usando el loc:
#Extraed todas las filas de la columna "price". 
df.loc[:,"price"]
print(df.loc[:,"price"])
# ¿De qué tipo es el resultado obtenido?
print(f'El tipo de resultado obtenido es: {type(df.loc[:,"price"])}')
```

```python
#Extraed todas las filas de las columnas "price" y "retail_price". ¿De qué tipo es el resultado obtenido?
df3 = df.loc[:,("price","retail_price")]
print(df3)
# ¿De qué tipo es el resultado obtenido?
print(f'El tipo de resultado obtenido es: {type(df3)}')
```

```python
#Extraed las filas pares, de las columnas 
# "rating", "rating_count", "rating_five_count", "rating_four_count", "rating_three_count", "rating_two_count", "rating_one_count"

df_pares = df.loc[::2,("rating", "rating_count", "rating_five_count", "rating_four_count", "rating_three_count", "rating_two_count", "rating_one_count")]
print(df_pares)
```

```python
#Extraer las filas cuyo precio sea mayor que 10 y sus unidades vendidas ("units_sold") sea mayor que 5000.
#Para ello hacemos mascaras
mascara = ((df["price"] > 10) & (df["units_sold"] > 5000))
print(mascara)
```
