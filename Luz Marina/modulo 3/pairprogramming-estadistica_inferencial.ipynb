2. Ejercicios intervalo de confianza: Para cada uno de los ejercicios interpreta los resultados y piensa que ventaja a nivel de negocio podemos sacar de calcularlo.

- Trabajas para una plataforma de comercio en línea y deseas determinar el intervalo de confianza al 95% para el precio promedio de productos con una cantidad de media de valoraciones al producto (mean_product_ratings_count) superior a 500.



```python
df_rating_mayor_500 = df[df['mean_product_ratings_count'] > 500]
media_price = df_rating_mayor_500['price'].mean() #estimacion puntual
error_rating = stats.sem(df_rating_mayor_500['price']) #eroor estandar
nivel_confianza = 0.95
valor_critico = stats.t.ppf((1 + nivel_confianza) / 2, df = len(df_rating_mayor_500['price'])-1)
valor_critico
limite_inferior_price = media_price - valor_critico * error_rating
limite_superior_price = media_price + valor_critico * error_rating
print(f'El intervalo de confianza del 95% para la variable price es:({limite_inferior_price}, {limite_superior_price})')

# El precio de los productos estara en un 95% de los casos entre 8.01 y 8.56 para los productos que tienen una valoracion media superior a 500 
```

- Eres un analista de ventas en una plataforma de comercio electrónico y quieres calcular el intervalo de confianza al 90% para el número promedio de unidades vendidas(mean_units_sold_per_product) por producto.


```python
media_unit_sold = df['mean_units_sold_per_product'].mean()#estimacion puntual

error = stats.sem(df['mean_units_sold_per_product']) #error estandar
nivel_confianza = 0.90
valor_critico = stats.t.ppf((1 + nivel_confianza) / 2, df = len(df['mean_units_sold_per_product'])-1)
valor_critico
limite_inferior_unit_sold = media_unit_sold  - valor_critico * error
limite_superior_unit_sold = media_unit_sold  + valor_critico * error
print(f'El intervalo de confianza del 90% para la variable unit sold es:({limite_inferior_unit_sold}, {limite_superior_unit_sold})')

# Las unidades vendidas estan en un 90% de probabilidad entre 4005.05 y 4672.96
```

- Trabajas en la gestión de inventario de una tienda en línea y necesitas calcular el intervalo de confianza al 99% para el precio promedio de productos(mean_product_prices) según su país de origen(origin_country).


```python
lista_paises = ['CN', 'US', 'VE', 'AT', 'SG', 'GB']
for country in lista_paises:
    df_pais = df[df['origin_country'] == country]
    media_product_price_country = df_pais['mean_product_prices'].mean()
    error = stats.sem(df['mean_product_prices']) #error estandar
    nivel_confianza = 0.99
    valor_critico = stats.t.ppf((1 + nivel_confianza) / 2, df = len(df_pais['mean_product_prices'])-1)
    valor_critico
    limite_inferior = media_product_price_country  - valor_critico * error
    limite_superior = media_product_price_country  + valor_critico * error
    print(f'El intervalo de confianza del 99% para la variable mean product price y el pais {country} es:({limite_inferior}, {limite_superior})')
```

- Eres un gerente de ventas en una plataforma de comercio electrónico y quieres determinar el intervalo de confianza al 95% para la valoración promedio de los vendedores(rating_employee) según si ofrecen envío express o no(shipping_is_expres).


```python
# CHAT GPT (chuletilla para funcion cuando hay varios grupos distintos)

def calcular_intervalo_confianza(grupo):
    # Tamaño de la muestra
    n = len(grupo)
    
    # Media de la muestra
    mean = np.mean(grupo)
    
    # Desviación estándar de la muestra
    std_err = stats.sem(grupo)  # Error estándar
    
    # Intervalo de confianza al 99%
    h = std_err * stats.t.ppf((1 + 0.99) / 2., n-1)
    
    return mean, mean - h, mean + h

# Agrupar por país de origen y calcular el intervalo de confianza
intervalos = df.groupby('origin_country')['mean_product_prices'].apply(calcular_intervalo_confianza)

# Convertir a DataFrame para facilitar la visualización
intervalos_df = pd.DataFrame(intervalos.tolist(), index=intervalos.index, columns=['Media', 'Límite Inferior', 'Límite Superior'])

print(intervalos_df)
```
